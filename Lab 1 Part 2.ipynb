```python
#load class
import numpy as np
import pandas as pd # in case it's not installed then install using conda create -c conda-forge -n name_of_my_env python pandas
# more installations information can be found: https://pandas.pydata.org/docs/getting_started/install.html
import os
#load sklearn module for creating and evaluating ML models. In case sklearn isn't installed,
#follow this link https://scikit-learn.org/stable/install.html for further details on the installation process.
from sklearn.neighbors import KNeighborsClassifier #load your classifier. In this code, I have used KNN. You can choose other algorithms. You have to use at least 3 to complete this.
from sklearn.preprocessing import StandardScaler #module for perform scaling
from sklearn.model_selection import train_test_split #module for splitting datatset
from sklearn import metrics #module for evaluating performance
```

```python
#load your data
df = pd.read_csv("data.csv") #change the name accordingly
df.head() # prints top 5 rows from the datatset to check data is load or no
```

```python
# Check for missing values
missing_values = df.isnull().sum()
print("Missing Values:\n", missing_values)
```

```python
# remove duplicatesd
df = df.drop_duplicates()
```

```python
# prepare features
x = df.drop(['Label'],axis=1) #remove class or label
y = df['Label'] #load label
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=.2) #split datatset. Here ratio is 80:20. Change accordingl
```

```python
#Scale the data using standardization
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train) #scale training set
x_test = scaler.transform(x_test) #scale test set
```

```python
z = KNeighborsClassifier(n_neighbors=3) # KNN classifier for 3 neighbours
KNN = z.fit(x_train,y_train) # start training
predict = KNN.predict(x_test) # performance in the test set
```

```python
print("Accuracy:", metrics.accuracy_score(y_test,predict)) # evaluating the performance based on accuracy
```

```python
# library for save and load scikit-learn models
import pickle
# file name, recommending *.pickle as a file extension
filename = "model.pickle"
# save model
pickle.dump(z, open(filename, "wb"))
```

```python
# 1st Algorithm: Decision Tree
from sklearn.tree import DecisionTreeClassifier

tree = DecisionTreeClassifier(random_state=42)

tree.fit(x_train, y_train)

tree_predict = tree.predict(x_test)

tree_accuracy = metrics.accuracy_score(y_test, tree_predict)
tree_precision = metrics.precision_score(y_test, tree_predict, zero_division=0)
tree_recall = metrics.recall_score(y_test, tree_predict, zero_division=0)
tree_f1 = metrics.f1_score(y_test, tree_predict, zero_division=0)
```

```python
#2nd Algorithm: Random Forest
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=100, random_state=42)

rf.fit(x_train, y_train)

rf_predict = rf.predict(x_test)

rf_accuracy = metrics.accuracy_score(y_test, rf_predict)
rf_precision = metrics.precision_score(y_test, rf_predict, zero_division=0)
rf_recall = metrics.recall_score(y_test, rf_predict, zero_division=0)
rf_f1 = metrics.f1_score(y_test, rf_predict, zero_division=0)
```

```python
#3rd Algorithm: Logistic Regression
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=1000)

lr.fit(x_train, y_train)

lr_predict = lr.predict(x_test)

lr_accuracy = metrics.accuracy_score(y_test, lr_predict)
lr_precision = metrics.precision_score(y_test, lr_predict, zero_division=0)
lr_recall = metrics.recall_score(y_test, lr_predict, zero_division=0)
lr_f1 = metrics.f1_score(y_test, lr_predict, zero_division=0)
```

```python
#Add in the the performance of KNN
knn_accuracy = metrics.accuracy_score(y_test, predict)
knn_precision = metrics.precision_score(y_test, predict, zero_division=0)
knn_recall = metrics.recall_score(y_test, predict, zero_division=0)
knn_f1 = metrics.f1_score(y_test, predict, zero_division=0)

#Create performance table
performance_table = pd.DataFrame({
    "Algorithm": ["KNN", "Decision Tree", "Random Forest", "Logistic Regression"],
    "Accuracy": [knn_accuracy, tree_accuracy, rf_accuracy, lr_accuracy],
    "Precision": [knn_precision, tree_precision, rf_precision, lr_precision],
    "Recall": [knn_recall, tree_recall, rf_recall, lr_recall],
    "F1 Score": [knn_f1, tree_f1, rf_f1, lr_f1]
})

print(performance_table)
```

```python
#Saving best algorithm
#Due to each model having a score of 1 in each catagory, there is not a singular best model
#Though we are choosing to proceed with the Random Forest algorithm. Please refer to the Lab Report for further reasoning.
pickle.dump(rf, open(filename, "wb"))
```
